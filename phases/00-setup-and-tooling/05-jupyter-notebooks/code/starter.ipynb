# Jupyter 快速入门

## 5 分钟学会 notebook

这个 notebook 里每个块叫「单元格」（cell）。
- 这种灰底文字是 **Markdown 单元格**，用来写笔记
- 下面白色的是 **代码单元格**，用来写 Python

**先学一个键：`Shift + Enter` = 运行当前单元格**

```python
# Shift+Enter 运行这个单元格试试
import numpy as np
import matplotlib.pyplot as plt

print("Hello Jupyter!")
print(f"NumPy version: {np.__version__}")
```

## 1. 代码和输出在一起

跑下面这行，结果直接嵌在代码下面。和 .py 脚本最大的区别就在这里。

```python
data = np.random.randn(1000)
data.mean(), data.std()
```

## 2. 变量跨单元格持久存在

上面定义了 `data`，下面直接用。不重新跑。

```python
plt.hist(data, bins=30, edgecolor='black')
plt.title('1000 个正态分布随机数')
plt.show()
```

## 3. 魔术命令

以 `%` 开头，不是 Python，是 Jupyter 特有的。

```python
# %timeit 多次运行取平均
%timeit np.random.randn(10000)
```

```python
# %%time 只跑一次
%%time
big = np.random.randn(5000, 5000)
result = big @ big.T
```

```python
# %who 看当前有哪些变量
%who
```

## 4. DataFrame 的富文本显示

不用 `print()`，直接写变量名就能出表格。

```python
import pandas as pd

df = pd.DataFrame({
    "模型": ["线性回归", "随机森林", "神经网络", "XGBoost"],
    "准确率": [0.72, 0.89, 0.94, 0.91],
    "训练时间(s)": [0.1, 2.3, 45.6, 8.2]
})
df  # 直接写变量名，自动渲染成表格
```

## 5. 关键快捷键

| 按键 | 作用 |
|------|------|
| `Shift+Enter` | 运行当前单元格 |
| `Esc` | 进入命令模式（左边变蓝） |
| `Enter` | 进入编辑模式（左边变绿） |
| `A` | 在上方插入单元格（命令模式） |
| `B` | 在下方插入单元格（命令模式） |
| `DD` | 删除单元格（命令模式） |
| `M` | 转为 Markdown 单元格 |
| `Y` | 转为代码单元格 |

**先记住 `Shift+Enter` 就够了，其他用到再查。**
